# Phase 3: Data Preprocessing

```python
df = pd.DataFrame(sample_data)
```

```python
df = pd.DataFrame(sample_data)

print(df)
```

```python
# Check missing values

print(df.isnull().sum())
```

```python
# Define valid amino acids

VALID_AMINO_ACIDS = set('ACDEFGHIKLMNPQRSTVWY')

print(VALID_AMINO_ACIDS)
```

```python
# Function to clean protein sequences

def clean_sequence(seq):

    # Check if input is actually text/string
    if not isinstance(seq, str):
        return None

    # Convert lowercase to uppercase
    seq = seq.upper()

    # Keep ONLY valid amino acids
    seq = ''.join(c for c in seq if c in VALID_AMINO_ACIDS)

    # Remove sequences shorter than 10 amino acids
    return seq if len(seq) >= 10 else None
```

```python
# Apply cleaning function to dataset

df['cleaned_sequence'] = df['sequence'].apply(clean_sequence)

# Show original vs cleaned sequences
print(df[['sequence', 'cleaned_sequence']])
```

```python
# Remove rows where cleaning returned None

df = df.dropna(subset=['cleaned_sequence'])

print("Remaining proteins after cleaning:")
print(len(df))

print("\nCleaned Dataset:")
print(df[['protein_id', 'cleaned_sequence']])
```

```python
# Function to split GO terms

def parse_go_terms(go_string):

    # Handle missing values
    if pd.isna(go_string):
        return []

    # Split labels by comma
    terms = [t.strip() for t in go_string.split(',')]

    return terms
```

```python
# Apply GO term parsing

df['go_terms_list'] = df['go_terms'].apply(parse_go_terms)

# Show parsed GO labels
print(df[['go_terms', 'go_terms_list']])
```

```python
from sklearn.preprocessing import MultiLabelBinarizer

# Create encoder
mlb = MultiLabelBinarizer()

# Convert GO term lists into binary matrix
Y = mlb.fit_transform(df['go_terms_list'])

print("Label Matrix Shape:")
print(Y.shape)

print("\nGO Term Classes:")
print(mlb.classes_)

print("\nEncoded Labels:")
print(Y)
```

```python
# Calculate sequence lengths

df['seq_length'] = df['cleaned_sequence'].apply(len)

print(df[['cleaned_sequence', 'seq_length']])
```

```python
# === FILTER LONG SEQUENCES ===

# Keep only proteins with length <= 1000
df = df[df['seq_length'] <= 1000]

print("After Length Filtering:")
print(len(df))


# === TRAIN / TEST SPLIT ===

from sklearn.model_selection import train_test_split

# Convert sequences column to Python list
sequences = df['cleaned_sequence'].tolist()

# Split dataset
X_train, X_test, Y_train, Y_test = train_test_split(
    sequences,
    Y,
    test_size=0.2,
    random_state=42
)

print("\nTrain Size:")
print(len(X_train))

print("\nTest Size:")
print(len(X_test))

print("\nTraining Label Shape:")
print(Y_train.shape)


# === SAVE PREPROCESSED DATA ===

import os
import pickle

# Create processed data folder
save_dir = '/content/protein-function-prediction/data/processed'

os.makedirs(save_dir, exist_ok=True)

# Save train sequences
pd.DataFrame({
    'sequence': X_train
}).to_csv(f'{save_dir}/train_sequences.csv', index=False)

# Save test sequences
pd.DataFrame({
    'sequence': X_test
}).to_csv(f'{save_dir}/test_sequences.csv', index=False)

# Save label arrays
np.save(f'{save_dir}/Y_train.npy', Y_train)
np.save(f'{save_dir}/Y_test.npy', Y_test)

# Save MultiLabelBinarizer encoder
with open(f'{save_dir}/mlb_encoder.pkl', 'wb') as f:
    pickle.dump(mlb, f)

print("\nAll processed files saved successfully!")

print("\nSaved Files:")
print(os.listdir(save_dir))
```

# Phase 4: Sequence Encoding & Embeddings


```python
# === PHASE 4: SEQUENCE ENCODING ===

import numpy as np
import torch
import torch.nn as nn

print("Libraries loaded successfully!")
```

```python
# ── Method 1: Integer Encoding ──────────────────────────────

AMINO_ACIDS = 'ACDEFGHIKLMNPQRSTVWY'

aa_to_int = {aa: idx + 1 for idx, aa in enumerate(AMINO_ACIDS)}

# Special tokens
aa_to_int['PAD'] = 0
aa_to_int['UNK'] = 21

print(aa_to_int)
```

```python
def integer_encode(sequence, max_length=512):

    encoded = [aa_to_int.get(aa, 21) for aa in sequence]

    encoded = encoded[:max_length]

    padding_length = max_length - len(encoded)

    encoded = encoded + [0] * padding_length

    return encoded


seq = "MKTAY"

encoded = integer_encode(seq, max_length=10)

print(f"Sequence: {seq}")
print(f"Encoded: {encoded}")
```

```python
def one_hot_encode(sequence, max_length=100):

    vocab_size = len(AMINO_ACIDS)

    matrix = np.zeros((max_length, vocab_size))

    for i, aa in enumerate(sequence[:max_length]):

        if aa in aa_to_int and aa_to_int[aa] <= 20:

            col = aa_to_int[aa] - 1

            matrix[i, col] = 1.0

    return matrix


one_hot = one_hot_encode("MKTAY", max_length=5)

print("One Hot Shape:")
print(one_hot.shape)

print(one_hot)
```

```python
# === PHASE 5: BUILDING THE MODEL ===

import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
import numpy as np

print("PyTorch Loaded Successfully!")
print("Torch Version:", torch.__version__)
```

```python
# ── Custom Dataset Class ─────────────────────────────────────

class ProteinDataset(Dataset):

    def __init__(self, sequences, labels, max_length=512):

        self.sequences = sequences
        self.labels = labels
        self.max_length = max_length

        self.vocab = {
            aa: idx + 1
            for idx, aa in enumerate('ACDEFGHIKLMNPQRSTVWY')
        }

        self.vocab['PAD'] = 0
        self.vocab['UNK'] = 21

    def __len__(self):

        return len(self.sequences)

    def __getitem__(self, idx):

        seq = self.sequences[idx]

        encoded = [
            self.vocab.get(aa, 21)
            for aa in seq[:self.max_length]
        ]

        pad_len = self.max_length - len(encoded)

        encoded = encoded + [0] * pad_len

        x = torch.tensor(encoded, dtype=torch.long)

        y = torch.tensor(self.labels[idx], dtype=torch.float32)

        return x, y


print("ProteinDataset class created successfully!")
```

```python
# ── Model Architecture ───────────────────────────────────────

class ProteinFunctionPredictor(nn.Module):

    def __init__(
        self,
        vocab_size=22,
        embed_dim=128,
        hidden_dim=256,
        num_layers=2,
        num_labels=100,
        dropout=0.3
    ):

        super(ProteinFunctionPredictor, self).__init__()

        # Embedding Layer
        self.embedding = nn.Embedding(
            num_embeddings=vocab_size,
            embedding_dim=embed_dim,
            padding_idx=0
        )

        # BiLSTM Layer
        self.bilstm = nn.LSTM(
            input_size=embed_dim,
            hidden_size=hidden_dim,
            num_layers=num_layers,
            batch_first=True,
            bidirectional=True,
            dropout=dropout if num_layers > 1 else 0
        )

        lstm_output_dim = hidden_dim * 2

        # Self Attention
        self.attention = nn.MultiheadAttention(
            embed_dim=lstm_output_dim,
            num_heads=8,
            dropout=dropout,
            batch_first=True
        )

        # Batch Normalization
        self.batch_norm = nn.BatchNorm1d(lstm_output_dim)

        # Classification Head
        self.classifier = nn.Sequential(

            nn.Linear(lstm_output_dim, 512),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(512, 256),

            nn.ReLU(),

            nn.Dropout(dropout),

            nn.Linear(256, num_labels)

        )

    def forward(self, x):

        # Embedding
        embedded = self.embedding(x)

        # BiLSTM
        lstm_out, (hidden, cell) = self.bilstm(embedded)

        # Self Attention
        attn_out, _ = self.attention(
            lstm_out,
            lstm_out,
            lstm_out
        )

        # Mean Pooling
        pooled = attn_out.mean(dim=1)

        # Batch Norm
        pooled = self.batch_norm(pooled)

        # Classification
        output = self.classifier(pooled)

        return output


print("ProteinFunctionPredictor class created successfully!")
```
